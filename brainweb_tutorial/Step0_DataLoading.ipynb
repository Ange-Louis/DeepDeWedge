# Step 0: Data Loading (BrainWeb)


We apply DeepDeWedge to a simulated tomogram pair generated from the [BrainWeb](https://brainweb.bic.mni.mcgill.ca/) phantom. Unlike the original tutorial (which uses a real cryo-ET tilt series from the Tomo110 dataset), here the two "half" volumes ${\color{red}\mathbf{v}^0}$ and ${\color{blue}\mathbf{v}^1}$ are two independent reconstructions of the same BrainWeb ground-truth volume, each simulated with a missing wedge of $50°$.

**Note:** the data is assumed to already be available locally in `./brainweb_tutorial/brainweb_dataset/` — there is no download step here. Adjust the file names below (and in `brainweb_config.yaml`) to match the files you place in that folder.

```python
from ddw.utils.mrctools import load_mrc_data
from ddw.utils.visualization import plot_tomo_slices
```

We assume the following files are already present in `./brainweb_tutorial/brainweb_dataset/`:
- `tomo_even_frames.mrc` and `tomo_odd_frames.mrc`: the two missing-wedge-degraded reconstructions of the BrainWeb volume (used as $\mathbf{v}^0$ / $\mathbf{v}^1$ for model fitting).
- `tomo_all_frames.mrc`: an optional full/reference reconstruction (e.g. the noise-free BrainWeb ground truth, or a reconstruction combining both halves) used only here and in Step 3 for visual comparison.

If your files are named differently, either rename them to match, or edit the paths used throughout these notebooks and in `brainweb_config.yaml`.

Let's have a look at the reference/full BrainWeb volume:

```python
tomo_full = load_mrc_data("./brainweb_tutorial/brainweb_dataset/tomo_all_frames.mrc")
plot_tomo_slices(
    tomo_full.clamp(-3 * tomo_full.std(), 3 * tomo_full.std()), figsize=(10, 20)
).show()
```

And the two missing-wedge-degraded halves that will be used for model fitting:

```python
tomo_even = load_mrc_data("./brainweb_tutorial/brainweb_dataset/tomo_even_frames.mrc")
plot_tomo_slices(
    tomo_even.clamp(-3 * tomo_even.std(), 3 * tomo_even.std()), figsize=(10, 20)
).show()

tomo_odd = load_mrc_data("./brainweb_tutorial/brainweb_dataset/tomo_odd_frames.mrc")
plot_tomo_slices(
    tomo_odd.clamp(-3 * tomo_odd.std(), 3 * tomo_odd.std()), figsize=(10, 20)
).show()
```
